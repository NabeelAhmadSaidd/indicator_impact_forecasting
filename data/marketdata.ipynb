```python
import yfinance as yf
import pandas as pd

assets = {
    "AAPL": "Apple",
    "MSFT": "Microsoft",
    "^GSPC": "SP500",
    "^IXIC": "NASDAQ",
    "EURUSD=X": "EURUSD",
    "USDJPY=X": "USDJPY",
    "GC=F": "Gold",
    "CL=F": "CrudeOil",
    "BTC-USD": "Bitcoin",
    "ETH-USD": "Ethereum"
}

for symbol, name in assets.items():
    print(f"Downloading data for {name} ({symbol})...")
    df = yf.download(symbol, start="2010-01-01", end="2026-01-01", interval="1d")
    file_name = f"{name}_daily.csv"
    df.to_csv(file_name)
    print(f"Saved {file_name}")
```

GET AAPL 60M data (Just for Testing)

```python
import requests
import os
from datetime import datetime, timedelta

symbol = "AAPL"
resolution = "60"
start = datetime(2020,1,1)
end = datetime(2026,1,1)

current = start

while current < end:
    next_year = min(current + timedelta(days=365), end)

    url = f"https://api.marketdata.app/v1/stocks/candles/{resolution}/{symbol}"
    params = {
        "from": current.strftime("%Y-%m-%d"),
        "to": next_year.strftime("%Y-%m-%d"),
        "format": "csv"
    }

    r = requests.get(url, params=params)

    fname = f"{symbol}/{resolution}m/{symbol}_{resolution}m_{current.year}.csv"
    os.makedirs(os.path.dirname(fname), exist_ok=True)
    open(fname,"wb").write(r.content)

    print("saved", fname)

    current = next_year
```

```python
import pandas as pd
import os

def combine_aapl_60m_data():
    dfs = []
    for year in range(2020, 2026):
        fname = f"AAPL/60m/AAPL_60m_{year}.csv"
        if os.path.exists(fname):
            df = pd.read_csv(fname)
            dfs.append(df)
        else:
            print(f"File {fname} not found")
    
    if dfs:
        combined = pd.concat(dfs, ignore_index=True)
        combined.sort_values('t', inplace=True)
        combined.to_csv("AAPL_60m_2020-2025.csv", index=False)
        print("Combined CSV saved as AAPL_60m_2020-2025.csv")
    else:
        print("No files found to combine")

# Call the function
combine_aapl_60m_data()
```

```python
import pandas as pd

def convert_timestamp_to_datetime(csv_file):
    df = pd.read_csv(csv_file)
    df['t'] = pd.to_datetime(df['t'], unit='s')
    df.to_csv(csv_file, index=False)
    print(f"Converted timestamps in {csv_file} to datetime")

# Call the function
convert_timestamp_to_datetime("AAPL_60m_2020-2025.csv")
```
